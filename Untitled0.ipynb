```python
# EXPERIMENT 10: EXPECTATION MAXIMIZATION ALGORITHM

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.stats import norm

# STEP 1: CREATE TRAINING DATASET

np.random.seed(42)

data1 = np.random.normal(20, 3, 50)
data2 = np.random.normal(50, 4, 50)
data3 = np.random.normal(80, 3, 50)

X = np.concatenate([data1, data2, data3])

df = pd.DataFrame({
    'Data Value': X
})

print("========== TRAINING DATASET ==========")
display(df.head(15))

print("\nTotal Samples:", len(X))

# STEP 2: INITIALIZE PARAMETERS

K = 3
N = len(X)

means = np.percentile(X, [20, 50, 80])
variances = np.ones(K) * np.var(X)
weights = np.ones(K) / K

max_iterations = 100
tolerance = 1e-5

log_likelihood_history = []

print("\n========== INITIAL PARAMETERS ==========")
print("Means:", means)
print("Variances:", variances)
print("Weights:", weights)

# STEP 3: EXPECTATION MAXIMIZATION ALGORITHM

for iteration in range(max_iterations):

    # E-STEP: Calculate Responsibilities

    responsibilities = np.zeros((N, K))

    for k in range(K):
        responsibilities[:, k] = (
            weights[k] *
            norm.pdf(X, means[k], np.sqrt(variances[k]))
        )

    total = responsibilities.sum(axis=1, keepdims=True)
    responsibilities /= np.maximum(total, 1e-300)

    # M-STEP: Update Parameters

    Nk = responsibilities.sum(axis=0)

    new_means = np.sum(
        responsibilities * X[:, None],
        axis=0
    ) / Nk

    new_variances = np.sum(
        responsibilities * (X[:, None] - new_means)**2,
        axis=0
    ) / Nk

    new_weights = Nk / N

    # Calculate Log Likelihood

    likelihood = np.sum(
        np.log(np.maximum(total.flatten(), 1e-300))
    )

    log_likelihood_history.append(likelihood)

    # Check Convergence

    if iteration > 0:
        if abs(
            log_likelihood_history[-1] -
            log_likelihood_history[-2]
        ) < tolerance:
            means = new_means
            variances = new_variances
            weights = new_weights
            break

    means = new_means
    variances = new_variances
    weights = new_weights

# STEP 4: FINAL CLUSTER ASSIGNMENT

cluster_labels = np.argmax(responsibilities, axis=1)

print("\n========== FINAL EM PARAMETERS ==========")

for k in range(K):
    print("\nCluster:", k + 1)
    print("Mean:", round(means[k], 2))
    print("Variance:", round(variances[k], 2))
    print("Weight:", round(weights[k], 4))

print("\nTotal Iterations:", len(log_likelihood_history))

# STEP 5: DISPLAY PREDICTION RESULTS

results = pd.DataFrame({
    'Data Value': X,
    'Cluster': cluster_labels + 1
})

print("\n========== CLUSTERING RESULTS ==========")
display(results.head(20))

# STEP 6: CLUSTER DISTRIBUTION BAR GRAPH

cluster_counts = pd.Series(
    cluster_labels + 1
).value_counts().sort_index()

plt.figure(figsize=(8, 5))

plt.bar(
    cluster_counts.index.astype(str),
    cluster_counts.values
)

plt.title("EM Algorithm - Cluster Distribution")
plt.xlabel("Clusters")
plt.ylabel("Number of Samples")
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()

# STEP 7: CLUSTER VISUALIZATION

plt.figure(figsize=(10, 5))

for k in range(K):
    cluster_data = X[cluster_labels == k]

    plt.scatter(
        cluster_data,
        np.zeros(len(cluster_data)) + k,
        label=f'Cluster {k + 1}'
    )

plt.title("EM Algorithm - Data Clustering")
plt.xlabel("Data Values")
plt.ylabel("Cluster Number")
plt.yticks(range(K), [f'Cluster {i+1}' for i in range(K)])
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()

# STEP 8: LOG LIKELIHOOD CONVERGENCE GRAPH

plt.figure(figsize=(9, 5))

plt.plot(
    range(1, len(log_likelihood_history) + 1),
    log_likelihood_history,
    marker='o'
)

plt.title("EM Algorithm - Log Likelihood Convergence")
plt.xlabel("Iterations")
plt.ylabel("Log Likelihood")
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()

# STEP 9: GAUSSIAN MIXTURE DISTRIBUTION

plt.figure(figsize=(10, 5))

plt.hist(
    X,
    bins=25,
    density=True,
    alpha=0.5,
    label='Training Data'
)

x_axis = np.linspace(X.min(), X.max(), 500)

mixture = np.zeros_like(x_axis)

for k in range(K):

    component = weights[k] * norm.pdf(
        x_axis,
        means[k],
        np.sqrt(variances[k])
    )

    mixture += component

    plt.plot(
        x_axis,
        component,
        linestyle='--',
        label=f'Cluster {k+1} Gaussian'
    )

plt.plot(
    x_axis,
    mixture,
    linewidth=2,
    label='Combined Gaussian Mixture'
)

plt.title("Gaussian Mixture Model using EM")
plt.xlabel("Data Values")
plt.ylabel("Probability Density")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()

# STEP 10: FINAL RESULT

print("\n========== FINAL RESULT ==========")

print("Expectation Maximization Algorithm")
print("Implemented Successfully.")

print("Number of Clusters:", K)
print("Total Iterations:", len(log_likelihood_history))

print("Final Cluster Means:", np.round(means, 2))

print("Log Likelihood:",
      round(log_likelihood_history[-1], 2))

print("\nEM Clustering Completed Successfully.")
```
